#### Check if any extracts are repeated

```python
def check_for_repeats(subj):
    df1 = pd.read_csv(f'./data/trials/{subj}/{subj}_vv_trials_modded.csv')
    df2 = pd.read_csv(f'./data/trials/{subj}/{subj}_va_trials_modded.csv')
    df3 = pd.read_csv(f'./data/trials/{subj}/{subj}_vva_trials_modded.csv')
    df = pd.concat([df1, df2, df3])

    # df_unique = df[df['Modality'] == 'vva']
    # df_unique

    de_combos = []
    for i, r in df.iterrows():
        if r['SpeakerDyad'] == r['ListenerDyad'] and r['SpeakerExtract'] == r['ListenerExtract']:
            de_combos.append((r['SpeakerDyad'], r['SpeakerExtract']))
        else:
            de_combos.append((r['SpeakerDyad'], r['SpeakerExtract']))
            de_combos.append((r['ListenerDyad'], r['ListenerExtract']))

    # print(de_combos)

    # print(len(de_combos))
    # print(len(set(de_combos)))
    return de_combos
```

```python
for i in range(1, 21):
    lsts = check_for_repeats(i)
    print(len(lsts))
    print(len(set(lsts)))
    print('--------------')
```
